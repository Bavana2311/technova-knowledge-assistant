```python
# Testing the FAISS retriever

from src.ingestion.Document_loader import load_documents
from src.ingestion.Chunker import chunk_document
from src.retrieval.faiss_retriever import get_retriever
import yaml
config = yaml.safe_load(open("./config.yaml"))

docs = load_documents("./data/documents/Employee_benefits.txt")

chunks = chunk_document(docs)

retriever = get_retriever(
    config,
    chunks_if_needed=chunks
)

query = "Attendance?"

results = retriever.similarity_search(query, k=5)

print("Number of retrieved chunks:", len(results))

for i, doc in enumerate(results, start=1):
    print(f"\n--- Result {i} ---")
    print(doc.page_content)
    print("Metadata:", doc.metadata)
```

```python
# Testing the FAISS retriever

from src.ingestion.Document_loader import load_documents
from src.ingestion.Chunker import chunk_document
from src.retrieval.faiss_retriever import create_retriever
import yaml

config = yaml.safe_load(open("./config.yaml"))

docs = load_documents("./data/documents/Employee_Leave_Policy.pdf")
chunks = chunk_document(docs)

print("Number of chunks created:", len(chunks))

faiss_store = create_retriever(
    chunks,
    provider="gemini",
    config=config
)

query = "How many annual leave days do employees get?"

results = faiss_store.similarity_search_with_score(query, k=5)

for i, (doc, score) in enumerate(results, start=1):
    print(f"\n--- Result {i} ---")
    print("FAISS Distance:", score)
    print(doc.page_content)
    print("Metadata:", doc.metadata)
```

```python
from src.ingestion.Document_loader import load_documents
from src.ingestion.Chunker import chunk_document
from src.retrieval.hybrid_retriever import HybridRetriever
from src.reranking.reranker import CrossEncoderReranker

docs = load_documents("./data/documents/Employee_Handbook.docx")
chunks = chunk_document(docs)

hybrid_retriever = HybridRetriever(
    chunks,
    config_path="./config.yaml"
)

reranker = CrossEncoderReranker(
    config_path="./config.yaml"
)

query = "What is the probation period?"

hybrid_results = hybrid_retriever.retrieve(query, top_k=5)

# Ignore the retrieval score because it is not used here
hybrid_docs = [doc for doc, _ in hybrid_results]

reranked_results = reranker.rerank(
    query,
    hybrid_docs,
    top_k=5
)

print(f"Query: {query}")
print(f"Number of reranked results: {len(reranked_results)}")

for i, (doc, score) in enumerate(reranked_results, 1):
    print(f"\n--- Reranked Result {i} ---")
    print("Reranker Score:", round(score, 4))
    print(doc.page_content)
```

```python
#testing the RAG Pipeline
import yaml

from src.ingestion.Document_loader import load_documents
from src.ingestion.Chunker import chunk_document

from src.retrieval.hybrid_retriever import HybridRetriever

from src.reranking.reranker import CrossEncoderReranker
from src.generation.llm_generator import GeminiGenerator

from src.pipeline.rag_pipeline import RAGPipeline


config = yaml.safe_load(open("./config.yaml"))

# Load documents
docs = load_documents(
    "./data/documents/Employee_Handbook.docx"
)

chunks = chunk_document(docs)

print("Number of chunks:", len(chunks))


# Hybrid (builds FAISS + BM25 internally from chunks)
hybrid = HybridRetriever(
    chunks,
    config_path="./config.yaml"
)


# Reranker
reranker = CrossEncoderReranker(
    config_path="./config.yaml"
)


# Gemini
generator = GeminiGenerator()


# Complete RAG
rag = RAGPipeline(
    hybrid_retriever=hybrid,
    reranker=reranker,
    generator=generator,
    session_id="test_session"
)


result = rag.query(
    "What is the probation period?"
)


print("\nANSWER:")
print(result["answer"])

print("\nSOURCES:")
for source in result["sources"]:
    print(source)
```
